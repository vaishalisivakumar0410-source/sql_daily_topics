```python
%%sql
create table emp(
    emp_id int primary key,
    name varchar(100) not null,
    age int  check(age > 18),
    manager_id int,
    foreign key(manager_id)
    references managers(mg_id)
    -- on delete cascade  
);
```

```python
%%sql
create table managers(
    mg_id int primary key,
    mg_name varchar(100) not null
);
```

```python
%%sql
select 
e.emp_id , 
e.name,
e.age, 
e.manager_id,
m.mg_name
from emp as e
join managers as m
on m.mg_id = e.manager_id;
```

```python
%%sql
select 
e.emp_id , 
e.name,
e.age, 
e.manager_id,
m.mg_name
from managers as m
left join emp as e
on m.mg_id = e.manager_id;
```

```python
%%sql
select 
e.emp_id , 
e.name,
e.age, 
e.manager_id,
m.mg_name
from emp as e
right join managers as m
on m.mg_id = e.manager_id;
```

```python
%%sql
select 
c.cust_id,
c.name,
c.number,
o.order_id,
o.order_amount
from customers as c
left join orders as o
on c.cust_id = o.cust_id;
```

```python
%%sql
select 
c.cust_id,
c.name,
c.number,
o.order_id,
o.order_amount
from customers as c
inner join orders as o
on c.cust_id = o.cust_id;
```
