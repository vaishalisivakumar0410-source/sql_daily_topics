```python
%%sql
select emp_id, emp_name,salary,
lag(salary,1,"New joinee")
over(
    partition by emp_id
    order by emp_id
)as prev_val
from employee_salary;
```

```python
%%sql
select emp_id, emp_name,salary,
lead(salary,1,"New joinee")
over(
    partition by emp_id
    order by emp_id
)as prev_val
from employee_salary;
```

```python
%%sql
select emp_id, emp_name, salary,
first_value(salary)
over(
    partition by emp_id
    order by salary desc
) as first_sal
from employee_salary;
```

```python
%%sql
select emp_id, emp_name, salary,
last_value(salary)
over(
    partition by emp_id
) as last_sal
from employee_salary;
```

```python
%%sql
select emp_id, emp_name, salary,
nth_value(salary, 8)
over(
    partition by emp_id
) as first_sal
from employee_salary;
```
