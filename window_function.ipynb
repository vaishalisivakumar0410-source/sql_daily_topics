```python
%%sql
select employee_id, employee_name, department,salary,job_title,
row_number()
over(
   order by employee_id
) as row_num
from employees;
```

```python
%%sql
select employee_id, employee_name, department,salary,job_title,
row_number()
over(
   partition by department
) as row_num
from employees;
```

```python
%%sql
select employee_id, employee_name, department,salary,job_title,
rank()
over(
    partition by department
   order by salary
) as row_num
from employees;
```

```python
%%sql
select employee_id, employee_name, department,salary,job_title,
dense_rank()
over(
    partition by department
   order by salary
) as row_num
from employees;
```
