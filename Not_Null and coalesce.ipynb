```python
%%sql
create table students (
    sid int primary key,
    name varchar(50),
    phone_number varchar(12),
    alternate_number varchar(12),
    parent_number varchar(12),
    email_id varchar(100),
    emg_number varchar(12)
);
```

```python
%%sql
select * from students;
```

```python
%%sql
select name , phone_number
from students;
```

```python
%%sql
select name , COALESCE(phone_number, alternate_number, parent_number,emg_number,email_id, 'company@gmail.com') as contact 
from students;
```
