```python
%%sql
create table order_1 (
    order_id int primary key,
    customer_name varchar(50),
    order_date DATE,
    order_time TIME,
    order_datetime DATETIME
);
```

```python
%%sql
insert into order_1 values (101,'John', '2025-07-01','09:30:15','2025-07-01 09:30:15'),(102, 'Sara', '2025-07-05', '14:15:20', '2025-07-05 14:15:20'),
(103, 'David', '2025-08-10', '18:45:30', '2025-08-10 18:45:30');
```

```python
%%sql
select current_date(), current_time(), now(), year('2025-07-05');
```

```python
%%sql
select customer_name, year(order_date),month(order_date), day(order_date),monthname(order_date) from order_1;
```

```python
%%sql
select order_id,customer_name, concat(day(order_date), '/', monthname(order_date), '/', year(order_date) ) as order_Date from order_1
```

```python
%%sql
select order_id,customer_name, concat(day(order_date), '/', monthname(order_date), '/', year(order_date) ) as order_Date,
concat(day(date_add(order_date, interval 10 day)), '/', monthname(date_add(order_date, interval 10 day)), '/', year(date_add(order_date, interval 10 day)) ) as due_Date
from order_1
```

```python
%%sql
select date_sub(current_date(), interval 10 day);
```

```python
%%sql
select order_id , customer_name, concat (day(order_date), '/', monthname(order_date), '/', year(order_date)) as new_date,
concat(day(date_sub(order_date, interval 10 day)) , '/', monthname(date_sub(order_date, interval 10 day)), '/', year(date_sub(order_date, interval 10 day)) ) as difference_day
from order_1;
```
