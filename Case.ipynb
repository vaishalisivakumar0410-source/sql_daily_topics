```python
%%sql
select employee_id, employee_name, department, salary,
case
when salary > 100000 then 'High'
else 'low'
end as sal_slab,
case 
when salary > 100000 and department= 'Technology' then salary * 0.50
else salary * 0.05
end as bonus
from employees;
```

```python
%%sql
select employee_id, employee_name, department, salary,
case
when salary > 100000 then 'High'
when salary > 50000 and salary <= 100000 then 'moderate'
else 'low'
end as sal_slab, 
```

```python
%%sql
select employee_id , employee_name , department , salary ,
case 
when salary > 10000 then salary * 0.50
else salary * 0.05
end as bonus,
case 
when (
    case 
     when salary > 100000 then salary * 0.10
     else salary * 0.05
     end
) > 10000 and department = 'Technology' then 'High bonus from Tech'
when(
    case
    when salary > 10000 then salary* 0.10
    else salary * 0.05
end
) < 10000 and department = 'Technology' then 'low bonus in Tech'
when (
    case 
    when salary > 100000 then salary  * 0.10
    else salary * 0.05
    end
) > 10000 and department not in ('Technology') then 'General High bonus'
else 'Low Bonus'
end as Bonus_label
from employees;
```

```python
%%sql
select sum(
    case 
        when salary > 100000 then salary * 0.50
        else salary * 0.05
        end
)
    as sum_bonus,
avg(
case 
when salary > 100000 then salary * 0.50
else salary * 0.05
end
)
as avg_bonus,

min(
    case 
    when salary > 100000 then salary * 0.50
    else salary * 0.05
    end
)
as min_bonus,

max(
    case 
    when salary > 10000 then salary * 0.50
    else salary * 0.05
    end
)
as max_bonus

from employees;
```
